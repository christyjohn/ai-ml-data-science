### Useful Jupyter tips. 

- Cells have 'Edit' mode and 'Command' mode.
- 'Enter' key functions as a linefeed where as 'Shift + Enter' executes the cell.
- You can enter 'Code' or 'Markdown' in the cells.
- Press 'Esc' to change from 'Edit' to 'Command' mode. Press 'Enter' to return back to 'Edit' mode.
- Press 'M' to change from 'Code' to Markdown, To change it back to 'Code' press 'y'.
- Preass 'Esc + b' to add a new cell below.'Esc + a' to add a cell above. 'Esc + d d' (double d) for deleting a cell.
- Adding images ```![description](image.jpg)```
- You can run terminal commands also from cells. e,g, ```ls```
